## Q6. Median value of horsepower

**Answer: Yes, it increased** (149.0 → 152.0)

```python
median_before = df['horsepower'].median()
median_before
```

```python
mode_value = df['horsepower'].mode()[0]
mode_value
```

```python
df['horsepower'] = df['horsepower'].fillna(mode_value)
```

```python
median_after = df['horsepower'].median()
median_after
```

```python
print(f'before: {median_before}, mode: {mode_value}, after: {median_after}')
```

## Q7. Sum of weights

**Answer: 0.51** (closest option to the actual value ~0.5188)

```python
asia = df[df['origin'] == 'Asia']
X = asia[['vehicle_weight', 'model_year']].iloc[:7].to_numpy()
X
```

```python
XTX = X.T.dot(X)
XTX
```

```python
XTX_inv = np.linalg.inv(XTX)
XTX_inv
```

```python
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
```

```python
w = XTX_inv.dot(X.T).dot(y)
w
```

```python
w.sum()
```
